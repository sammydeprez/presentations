# 1. Simple LLM Call

*Using Microsoft Semantic Kernel (Agent Framework)*

An LLM call is the simplest agent that is there. It can not execute any actions, it only returns tokens that fit to your prompt.

This notebook demonstrates the same concepts as the LangChain version, but using Microsoft Semantic Kernel.

```python
import semantic_kernel as sk
from semantic_kernel.connectors.ai.open_ai import AzureChatCompletion
from semantic_kernel.contents import ChatHistory
```

Define the kernel and model

```python
import os
from dotenv import load_dotenv

load_dotenv()

# Initialize the kernel
kernel = sk.Kernel()

# Add Azure OpenAI Chat Completion service
service_id = "chat-gpt"
kernel.add_service(
    AzureChatCompletion(
        service_id=service_id,
        deployment_name="gpt-4o-mini",
        endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
        api_key=os.getenv("AZURE_OPENAI_API_KEY"),
    )
)
```

Set up the chat history with system and user messages

```python
chat_history = ChatHistory()
chat_history.add_system_message(
    "You are a travel expert. Based on the user's input, suggest 1 travel destination."
)
chat_history.add_user_message("Cat")
```

Call the LLM/agent

```python
# Get the chat completion service
chat_service = kernel.get_service(service_id)

# Get the response
response = await chat_service.get_chat_message_content(
    chat_history=chat_history,
    settings=kernel.get_prompt_execution_settings_from_service_id(service_id),
)

# Add the response to chat history
chat_history.add_assistant_message(str(response))
```

Print result

```python
print(response)
```

Continue the conversation

```python
chat_history.add_user_message("I prefer warm places")

response = await chat_service.get_chat_message_content(
    chat_history=chat_history,
    settings=kernel.get_prompt_execution_settings_from_service_id(service_id),
)

chat_history.add_assistant_message(str(response))
print(response)
```
